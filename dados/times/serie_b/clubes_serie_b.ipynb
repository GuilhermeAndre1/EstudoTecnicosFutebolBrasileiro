# Clubes da Série B (2018–2025)
Coleta os clubes diretamente do site da CBF.

```python
import pandas as pd, requests, re, time
from bs4 import BeautifulSoup
from pathlib import Path

DIVISAO = "serie_b"
ANOS = list(range(2018, 2026))
BASE = f"https://www.cbf.com.br/futebol-brasileiro/tabelas/campeonato-brasileiro/serie-b"
UA = {"User-Agent":"Mozilla/5.0"}

linhas = []
for ano in ANOS:
    url = f"{BASE}/{ano}"
    print(">>>", url)
    r = requests.get(url, headers=UA, timeout=20)
    if r.status_code != 200: continue
    soup = BeautifulSoup(r.text, "html.parser")
    clubes = set()
    for a in soup.select("a"):
        txt = a.get_text(" ", strip=True)
        href = a.get("href") or ""
        if txt and "/times/campeonato-brasileiro/" in href:
            clubes.add(txt)
    for c in clubes:
        linhas.append({"ano":ano,"divisao":DIVISAO.upper(),"clube":c})
    time.sleep(1.5)

df = pd.DataFrame(linhas).drop_duplicates().sort_values(["ano","clube"])
out = f"clubes_{DIVISAO}_2018_2025.csv"
df.to_csv(out,index=False,encoding="utf-8")
df.head()
```
